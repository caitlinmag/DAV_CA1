# Getting list of all Cruelty free brands from the Peta website
- There are 7 pages of brands so I am checking each page for the brand name, there is a total of 6328 brand names on the list. 
- Using a list of the Alphabet to iterate through each page as they are sorted by letter. 
- Used the same method for the brands list that do animal test. 

```python
import time
import string
import pandas as pd
from selenium import webdriver
from bs4 import BeautifulSoup

driver = webdriver.Chrome()
cruelty_free_brands_list = []

alphabet = list(string.ascii_lowercase)

for letter in alphabet:    
    page_num = 1
    while True:
        if page_num == 1:
            url = f'https://crueltyfree.peta.org/companies-dont-test/?alpha={letter}'
        else:
            url = f'https://crueltyfree.peta.org/companies-dont-test/page/{page_num}/?alpha={letter}'
            
        driver.get(url)
        time.sleep(3)  

        soup = BeautifulSoup(driver.page_source, 'html.parser')
        
        num_brands_on_page = 0

        for brand in soup.select('ul.search-results li a'):
            name = brand.get_text(strip=True)
            if name:
                cruelty_free_brands_list.append(name)
                num_brands_on_page += 1
                
        if num_brands_on_page == 0:
            print (f"Finished for letter {letter}")
            break

        print(f"Letter: {letter}, Page: {page_num}), Total: {len(cruelty_free_brands_list)}")
        
        page_num += 1

driver.quit()

if cruelty_free_brands_list:
    df = pd.DataFrame(cruelty_free_brands_list, columns=['Brand Name'])
    df.to_csv('cruelty_free_brands.csv')
    print(f"Total Number of Brands {len(df)}")
```
