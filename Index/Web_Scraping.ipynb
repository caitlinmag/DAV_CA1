```python
from bs4 import BeautifulSoup
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
import time
```

```python
from selenium import webdriver
from selenium.webdriver import Chrome
```

```python

driver = Chrome()
cruelty_free_brands_list = []
for page_number in range (1, 8):
    driver.get('https://crueltyfree.peta.org/companies-dont-test/?alpha=&per_page=1000')
    time.sleep(6)
    soup = BeautifulSoup(driver.page_source, 'html.parser')
    page_count = 0
    for brand in soup.select('ul.search-results li a'):
        name = brand.get_text(strip=True)
        if name: 
            cruelty_free_brands_list.append(name)
            page_count += 1 
    print(f"{page_count}")

print("Total Number of Brands:")
print(len(cruelty_free_brands_list))
```
